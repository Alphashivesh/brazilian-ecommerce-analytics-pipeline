```python
import pandas as pd
```

```python
# Load the dataset
df = pd.read_csv('Olist_Sample_Exploration.csv')
```

```python
# showing first 5 rows
print("--- First 5 Rows ---")
display(df.head())
```

```python
# Check for missing values and data types
print("\n--- Dataset Info ---")
df.info()
```

```python
# Convert the purchase timestamp to a proper Datetime format
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
```

```python
# Handle missing product categories by labeling them as 'Unknown'
df['product_category_name'] = df['product_category_name'].fillna('Unknown')
```

```python
# Check for any duplicate rows and drop them
df = df.drop_duplicates()
```

```python
# Verify the cleaning worked
print("--- Cleaned Dataset Info ---")
df.info()
```

```python
# Calculate the total value of each item
df['total_item_value'] = df['product_price'] + df['shipping_cost']
```

```python
# Group the data by State to calculate Frequency (orders) and Monetary (revenue)
state_segmentation = df.groupby('customer_state').agg(
    total_orders=('order_id', 'nunique'),
    total_revenue=('total_item_value', 'sum'),
    avg_order_value=('total_item_value', 'mean')
).reset_index()
```

```python
# Sort by total revenue to find our "Champion" states
state_segmentation = state_segmentation.sort_values(by='total_revenue', ascending=False)
```

```python
# Round the numbers to 2 decimal places for cleaner reading
state_segmentation['total_revenue'] = state_segmentation['total_revenue'].round(2)
state_segmentation['avg_order_value'] = state_segmentation['avg_order_value'].round(2)
```

```python
# Display the Top 10 States
print("--- Top 10 Champion States by Revenue ---")
display(state_segmentation.head(10))
```

```python
# Export the cleaned master dataset to a new CSV
df.to_csv('Cleaned_Olist_Sample.csv', index=False)
print("Cleaned dataset successfully exported!")
```
